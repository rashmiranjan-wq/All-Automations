```python
import pandas as pd

query = """
WITH base AS (
    SELECT
        CAST(substr(t.ticket_creation_time, 1, 10) AS DATE) AS ticket_date,
        t.canonical_ticket_id,
        t.product_group,
        t.ticket_state,
        t.overall_last_agent_type_for_ticket,
        t.csat,
        t.genai_call_bot_disposition,
        t.genai_call_bot_sub_disposition
    FROM corp_analytics.cx_master_data_model_tbl t
    WHERE t.product_group = 'BBPS'
      AND t.source_channel = 'CHAT'
      AND t.overall_last_agent_type_for_ticket = 'GEN_AI_VOICE_BOT'
)

SELECT
    ticket_date,
    canonical_ticket_id,
    product_group,
    ticket_state,
    overall_last_agent_type_for_ticket,
    csat,
    genai_call_bot_disposition,
    genai_call_bot_sub_disposition
FROM base
WHERE
    (
        day_of_week(current_date) = 1
        AND ticket_date IN (
            current_date - INTERVAL '1' DAY,
            current_date - INTERVAL '2' DAY
        )
    )
    OR
    (
        day_of_week(current_date) <> 1
        AND ticket_date = current_date - INTERVAL '1' DAY
    )
ORDER BY ticket_date DESC
"""

result = get_sql_df(query)

if isinstance(result, tuple):
    df, code = result
else:
    df = result
    code = None

if df is not None and not df.empty:
    display(df)
else:
    print("No rows returned.", f"Code={code}" if code is not None else "")
```
