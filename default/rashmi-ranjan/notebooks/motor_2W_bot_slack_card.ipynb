```python
# Define the Google Sheet URL and the worksheet name
gsheet_url = "https://docs.google.com/spreadsheets/d/1jwqtHs6pJY2APFPunyZRHXyM0o9XSHrM4fTUmX-2Ph0/edit?gid=788581434#gid=788581434"
sheet_name = "Slack Card"

# Define the destination folder for the image
destination_folder = "data"

# Create the data directory if it doesn't exist
if not os.path.exists(destination_folder):
    os.makedirs(destination_folder)

# Initialize GoogleSheetReader with the provided sheet link and service account credentials
google_sheet_reader = GoogleSheetReader(gsheet_url, service_account)

# Download the specified range as an image
image_path = google_sheet_reader.download_google_sheet_as_image(destination_folder, sheet_name)

# Initialize SlackReportsTrigger with the Slack token
slack = SlackReportsTrigger(slack_token)

title = "Motor Insurance 2W Bot Audit Summary Snapshot"

# Secondary channel — same card, no CC tags
slack_channel_id_secondary = "C07K5MBQVDW"
message_without_cc = (
    "Please find below the summary of Motor Insurance 2W Bot audits done as on today"
)

slack.send_file_to_slack(
    slack_channel_id_secondary,
    image_path,
    title=title,
    message=message_without_cc,
)
```
